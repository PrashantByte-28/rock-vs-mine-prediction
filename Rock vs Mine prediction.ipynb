Importing Dependencies

```python
import numpy as np
import pandas as pd
from sklearn.preprocessing import LabelEncoder
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score
```

Data Collection and Data Processing

```python
sonar_data = pd.read_csv("sonar_data.csv", header = None)
```

```python
sonar_data.head()
```

Converting the data of last column labels

R --> 1

M --> 0

```python
encoder = LabelEncoder()

print(sonar_data[60].unique())

sonar_data[60] =  encoder.fit_transform(sonar_data[60])
```

```python
sonar_data.tail()
```

```python
sonar_data.info()
```

```python
sonar_data.shape
```

```python
sonar_data.describe()
```

```python
sonar_data.value_counts(60)
```

1 --> Rock

0 --> Mine

```python
sonar_data.groupby(60).mean()
```

```python
# Seperating Data and labels
X = sonar_data.drop(columns=60) # putting all the columns data in variable X except last column which 60
Y = sonar_data[60] # puuting the data of columns 60 in variable Y
```

```python
print(X)
print(Y)
```

Training and Testing Data

```python
X_train, X_test, Y_train, Y_test = train_test_split(X, Y, test_size = 0.1, stratify=Y, random_state=42) # test_size means how many percentage of data we are going to train from 100% we chose 0.1 then it is 10%
```

```python
print(X.shape,X_train.shape, X_test.shape)
```

Model Training --> Logistic Regression

```python
model = LogisticRegression()
```

```python
model.fit(X_train, Y_train)
```

Model Evaluation

```python
# Accuracy Training Data
X_train_prediction = model.predict(X_train)
training_data_accuracy = accuracy_score(X_train_prediction,Y_train)
```

```python
print("Accuracy on training data", training_data_accuracy)
```

```python
# Accuracy Test Data
X_test_prediction = model.predict(X_test)
test_data_accuracy = accuracy_score(X_test_prediction,Y_test)
```

```python
print("Accuracy on test Data", test_data_accuracy)
```
